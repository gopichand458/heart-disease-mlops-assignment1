## 1. Raw data and cleaning
UCI marks missing values with `?` (only `ca` and `thal`). Cleaning coerces types, binarises the target and drops duplicates. Imputation is done inside the model pipeline to avoid leakage.

```python
raw = load_raw()
print(raw.shape)
raw.head()
```

```python
df = clean(raw)
print(df.shape)
display(missing_report(df))
df.describe().T
```

## 2. Class balance

```python
display(Image(eda.class_balance(df)))
df['target'].value_counts(normalize=True)
```

## 3. Distributions

```python
display(Image(eda.numeric_histograms(df)))
display(Image(eda.numeric_boxplots(df)))
```

```python
display(Image(eda.categorical_counts(df)))
```

## 4. Correlations

```python
display(Image(eda.correlation_heatmap(df)))
df.corr()['target'].drop('target').sort_values(ascending=False)
```

## 5. Missing values

```python
display(Image(eda.missing_values(df)))
```
